```python
import matminer
import numpy as np
import pandas as pd
import pickle
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, precision_score, recall_score, f1_score, roc_auc_score
```

```python
df_sc = pd.read_csv("df_sc_ep_ElementProperty_Magpie_ftd.csv").dropna()
df_sc
```

```python
df_sc["having_tc"] = (df_sc["Tc"] >= 10).astype(int)
```

```python
X = df_sc.iloc[:, 3:-1]
y = df_sc['having_tc']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
X_train
```

```python
param_grid = {
    "n_estimators": [100, 200, 500],  # Number of trees in the forest
    "max_depth": [5, 10, 15],        # Maximum depth of individual trees
    "min_samples_split": [2, 5, 10],  # Minimum samples required to split a node
}

# param_grid = {
#     "n_estimators": [100, 200, 500, 1000],  # Increased range for number of trees
#     "max_depth": [5, 10],  # Lower range to potentially reduce overfitting
#     "min_samples_split": [2, 5, 10],
#     "min_samples_leaf": [1, 2, 5, 10],  # New parameter for leaf node samples
#     "max_features": ["auto", "sqrt", "log2"],  # New parameter for feature selection
#     "ccp_alpha": [0.0, 0.01, 0.1]  # New parameter for cost complexity pruning
# }
```

```python
# Create the Random Forest classifier
model = RandomForestClassifier()

# Create the GridSearchCV object
grid_search = GridSearchCV(estimator=model, param_grid=param_grid, cv=5, scoring="accuracy", n_jobs=-1, verbose=2)

# Fit the grid search to the training data
grid_search.fit(X_train, y_train)
```

```python
# Get the best model from the grid search
best_model = grid_search.best_estimator_

# Predict on the testing set
y_pred = best_model.predict(X_test)
```

```python
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)

# Confusion Matrix
conf_matrix = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:\n", conf_matrix)

# Precision
precision = precision_score(y_test, y_pred)
print("Precision:", precision)

# Recall
recall = recall_score(y_test, y_pred)
print("Recall:", recall)

# F1 Score
f1 = f1_score(y_test, y_pred)
print("F1 Score:", f1)

# ROC-AUC Score
roc_auc = roc_auc_score(y_test, y_pred)
print("ROC-AUC Score:", roc_auc)
```

```python
filename = 'sc_ep_rf_cl1.pkl'
with open(filename, 'wb') as f:
    pickle.dump(best_model, f)
```
